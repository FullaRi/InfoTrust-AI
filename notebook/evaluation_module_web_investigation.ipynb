```python
import pandas as pd
from sklearn.metrics import classification_report, confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt
```

## Load and Clean the Dataset:

```python
dataset = pd.read_csv("./dataset.csv")
```

```python
dataset.info()
```

```python
dataset = dataset[['text', "label (1= reel)", "prediction"]].copy()
dataset.dropna(axis=0, inplace=True)

dataset.info()
```

## SPLITTING

```python
X = dataset['text'].copy()
y_true = dataset['label (1= reel)'].copy()
y_pred = dataset['prediction'].copy()
```

## CLASSIFICATION REPORT

```python
c_report = classification_report(
    y_true,
    y_pred,
    target_names=["Fake", "Real"]
)

print("                   =============================")
print("                     RAPPORT DE CLASSIFICATION  ")
print("                   =============================")
print("\n")
print(c_report)

print("\n\n")
```

## CONFUSION MATRIX

```python
cm = confusion_matrix(y_true, y_pred)
ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["Fake", "Real"]).plot(cmap="Blues")
plt.title("Confusion Matrix - WEB INVESTIGATION")
plt.grid(False)
plt.show()
```
